---

## 1. The Clean Project Structure

Your `olist_docker` directory should look exactly like this on your Mac:

```text
olist_docker/                 <-- You open this folder in VS Code
├── .git/                     <-- Git history (Managed on Mac)
├── .env                      <-- API keys & Snowflake credentials (Managed on Mac)
├── Dockerfile                <-- The blueprint for the Factory
├── docker-compose.yml        <-- The power switch for the Factory
├── dagster_olist/            <-- Python ingestion scripts
├── dbt_olist_github/         <-- SQL transformation models
├── explore_olist/            <-- Jupyter Notebooks
└── venv_dbt_git/             <-- ONLY used for VS Code autocomplete (Lives on Mac)

```

## 2. Who Does What?

To stop the spaghetti, assign strict jobs to your three tools:

| Tool | Its Only Job | Where it Happens |
| --- | --- | --- |
| **MacBook Terminal & VS Code** | Writing code, saving files, and running `git add` / `git commit`. | On your host Mac. |
| **`venv_dbt_git`** | **Syntax Highlighting only.** You activate this in VS Code so your editor doesn't show red error lines under `import pandas`. You do *not* run heavy pipelines here anymore. | On your host Mac. |
| **Docker** | **Execution.** Running `dbt build`, orchestrating Dagster, and executing Jupyter cells. | Inside the container. |

---

## 3. Your New Daily Workflow

Here is exactly what you do when you sit down to work:

**Step 1: Turn on the Factory (Docker)**
Open your Mac terminal, navigate to `olist_docker`, and start the engines:

```bash
docker compose up -d

```

*Result: Dagster is now running at `localhost:3000` and Jupyter at `localhost:8888`. They will stay on in the background.*

**Step 2: Write Code (VS Code & Venv)**

* Open VS Code on your Mac.
* Set your Python Interpreter in VS Code to `venv_dbt_git` (so autocomplete works).
* Write a new SQL model in `dbt_olist_github/models/staging/` or edit an `.ipynb` file. When you hit save, the file instantly syncs into the running Docker container.

**Step 3: Test and Run (Inside Docker)**
Never run `dbt run` on your Mac terminal again. Instead, execute it *inside* the running container from your Mac terminal:

```bash
docker exec -it olist_dagster dbt run --project-dir dbt_olist_github

```

*Alternatively, just open Dagster at `localhost:3000` in your browser and click "Materialize" to let the container do the work.*

**Step 4: Save to History (Git on Mac)**
Once the code works perfectly inside Docker, use your Mac terminal to commit the changes:

```bash
git status
git add dbt_olist_github/models/staging/
git commit -m "feat(staging): build new logistics model"
git push origin main

```

**Step 5: Shut Down**
When you are done for the day, turn the factory off:

```bash
docker compose down

```
