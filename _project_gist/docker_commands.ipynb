### 1. Lifecycle & Running Services (Docker Compose)

These are the commands you will use daily to spin your "Factory" up and down.

* **`docker compose up -d`**
* **What it does:** Starts all the services defined in your `docker-compose.yml` file in the background (detached mode, indicated by `-d`).
* **When to use it:** When you sit down to work and want your Dagster and Jupyter servers running.


* **`docker compose up --build -d`**
* **What it does:** Rebuilds your container images from scratch (using your `Dockerfile` and `requirements.txt`) and then starts them in the background.
* **When to use it:** Whenever you modify your `requirements.txt` or add a new dependency (like we just did with `pyarrow`).


* **`docker compose down`**
* **What it does:** Gracefully stops and removes all running containers, networks, and default resources created by your compose file.
* **When to use it:** When you are done working for the day and want to clean up system resources.



---

### 2. Inspection & Debugging

When something isn't working as expected, these commands let you see what is happening inside the containers.

* **`docker ps`**
* **What it does:** Lists all currently **running** containers, showing their Container ID, image name, uptime, and mapped network ports.
* **When to use it:** To quickly verify if your services are alive. *(Tip: Use `docker ps -a` to see all containers, including stopped ones).*


* **`docker compose logs`**
* **What it does:** Streams the console output and error logs for all services in your compose file.
* **When to use it:** When a container fails to start or throws a runtime error. *(Tip: Target a specific service like `docker compose logs jupyter` or add `--tail=50` to see only the last 50 lines).*



---

### 3. Interacting with Running Containers

* **`docker exec -it <container_name> <command>`**
* **What it does:** Opens an interactive terminal session (`-it`) *inside* a running container to execute a command.
* **When to use it:** When you want to run CLI tools (like `dbt run` or a Python script) inside the standardized Linux container environment without leaving your terminal.
* *Example:* `docker exec -it olist_dagster dbt run --project-dir dbt_olist_github/dbt_olistgithub`



---

### 4. Cleanup & Maintenance

Over time, building and destroying images can accumulate unused data on your Mac. These commands keep your system lean.

* **`docker system prune -a`**
* **What it does:** Deletes all stopped containers, unused networks, and dangling images that are no longer associated with a running service.
* **When to use it:** When your Mac's storage is getting tight and you want to purge old Docker build caches. *(Warning: This will delete unused images, meaning the next build will need to re-download base layers).*
