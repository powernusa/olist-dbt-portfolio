```python
import io
import pandas as pd
import requests

url = "https://mydataaijournal.com/public/andy-data/olist-data/orders_milestone.parquet"

print("Fetching dataset from mydataaijournal.com...")
response = requests.get(url)
response.raise_for_status()

df_milestones = pd.read_parquet(io.BytesIO(response.content))
print(f"Successfully loaded {len(df_milestones):,} rows into memory!")
```

```python
df_milestones.head()
```

```python
df_milestones.columns = df_milestones.columns.str.lower()

df_milestones.head()
```

```python
df_milestones = df_milestones[df_milestones["milestone_name"]!="1_purchased"]

df_milestones.head()
```

```python
df_milestones.info()
```

```python
df_milestones = (
    df_milestones.dropna(subset=["hours_since_last_step"])
)

df_milestones.info()
```

```python



distribution_summary = (
    df_milestones.groupby("milestone_name")["hours_since_last_step"]
                 .describe(percentiles=[0.5,0.75,0.95,0.99])
)

distribution_summary
```

# 📊 Logistics: Delivery Bottleneck Analysis

> **Executive Summary:** The table breaks down the fulfillment pipeline to show exactly where delays happen. Instead of tracking the total time since checkout, it measures the distinct time spent between each step of the process.
<br><br>
* **Rows (The Stages):** Each row represents a specific, isolated phase of the fulfillment journey (e.g., payment approval wait time, warehouse packing time before carrier handoff, and actual transit time). <br> <br>
* **Columns (Percentile Time Spent):** The percentiles of hours elapsed since the previous step. They show the distribution of phase durations across all orders—ranging from typical duration (50%) to worst-case bottlenecks (99%).
