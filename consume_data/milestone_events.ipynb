```python
import io
import pandas as pd
import requests

url = "https://mydataaijournal.com/public/andy-data/olist-data/milestone_data.parquet"

print("Fetching dataset from mydataaijournal.com...")
response = requests.get(url)
response.raise_for_status()

df_milestones = pd.read_parquet(io.BytesIO(response.content))
print(f"Successfully loaded {len(df_milestones):,} rows into memory!")
```
